# 03 — Data Quality and Outliers: assignment

**Purpose.** Identify missing, malformed, duplicated, and extreme values before deciding how to prepare predictors. This combines the roadmap's Level 3 missing-value work with Level 4 outlier review.

**Prerequisites and scope.** Continue with data/raw/sample.xlsx and the column map used in notebook 01. Audit both the full 25,536-row source and the 24,142-row mature FPD cohort where relevant. Preserve the raw source. Do not build a model here.

**Learning objectives.** Separate application-key problems from duplicate feature rows; diagnose mixed types and missingness; assess extreme values without automatic deletion; choose defensible treatment rules that can later be fitted on training data only.

**Expected notebook-wide deliverables.** Quality audit tables, documented anomaly examples or counts, a missingness and outlier treatment plan, and a concise handoff to feature engineering. Write your own analysis and code below the assignment cells.

## Task 1 — Check record and key integrity

- [ ] Verify dimensions, source sheet, date range, and column meanings.
- [ ] Check missing and repeated application IDs, exact full-row duplicates, and duplicates after excluding the ID as distinct questions.
- [ ] Investigate any repeated ID and explain what a row represents before proposing deduplication.

**Acceptance criteria.** Report counts for each check and distinguish a repeated ID from identical application records. For the current source, reconcile 25,536 rows, 25,535 unique IDs, zero exact full-row duplicates, and the 15 repeated feature rows seen after excluding ID, or explain any difference.

**Deliverable.** Key-integrity table plus a documented decision on whether records should be retained.

## Task 2 — Map missingness

- [ ] Show missing count and percentage for every field in the full source and in the mature FPD cohort.
- [ ] Pay particular attention to employer_bin_gcvp, loan_term_months, client_id_document_date, and the credit-bureau fields.

**Acceptance criteria.** Percentages name their denominator; the table includes all 20 original fields or explicitly accounts for dropped columns. Missingness patterns and any difference between full and mature cohorts are explained. Do not interpret a missing value as zero by default.

**Deliverables.** Two missingness summaries, one co-missingness or blank-value check, and written observations.

## Task 3 — Validate types, values, and field meaning

- [ ] Identify date, categorical, numeric, ID, target, and maturity fields and their expected types.
- [ ] Inspect nonnumeric entries in active_loans_count_credit_bureau and any other supposedly numeric field.
- [ ] Check binary target/maturity values, date validity, negative counts or amounts, and suspicious category spelling or whitespace.
- [ ] Resolve the meaning of client_id_document_date using source headers or a data owner before deriving age. The alternative workbook labels this field Client_ID_Expiry_Date.

**Acceptance criteria.** Every issue has an observed count or explicit “none found,” plus a proposed disposition. No assumption that an ID-document date is a birth date is presented as fact.

**Deliverable.** A field-validation register with issue, evidence, and intended treatment.

## Task 4 — Review outliers and zero values

- [ ] Report minimum, median, high percentiles, and maximum for income, requested amount, outstanding balance, loan term, and key bureau counts.
- [ ] Count zero and negative values where they may make ratios invalid; the current source includes zero declared income.
- [ ] Compare at least two plausible outlier treatments for later models, such as retaining values, transforming skewed values, or capping extremes.

**Acceptance criteria.** Percentile cutoffs and units are stated; extreme values are reviewed rather than automatically removed. The rationale distinguishes the needs of a linear baseline from a tree-based model. Any proposed cap or transform is marked for fitting on training data only.

**Deliverables.** Outlier table or plots and a short treatment rationale.

## Task 5 — Test whether missingness may be informative

- [ ] Within the mature FPD cohort, compare outcome counts and rates for present versus missing employer BIN, loan term, and at least one bureau field.
- [ ] Note small positive counts and possible process explanations.
- [ ] Decide whether missing-indicator features are worth testing later.

**Acceptance criteria.** Every comparison includes group size and positive count; associations are described as descriptive, not causal. No outcome-based imputation values are chosen from the future validation or test windows.

**Deliverable.** Missingness-versus-outcome table and cautious interpretation.

## Task 6 — Specify a preprocessing contract

- [ ] Write a field-by-field plan for categorical missing values, numeric missing values, mixed types, and outliers.
- [ ] Explain which values should remain missing until the later modeling pipeline and how fitted statistics will use training rows only.
- [ ] State how UNKNOWN, numeric imputation, missing indicators, and any transforms would be compared or validated.
- [ ] Record columns that are never predictors: application ID, both targets, both maturity flags, and active_delinquency_days. Treat manager as a separate experiment.

**Acceptance criteria.** The plan avoids blanket fillna(0), preserves the raw workbook, names every affected field, and prevents validation/test data from determining imputation or clipping thresholds.

**Deliverable.** A concise preprocessing decision table.
